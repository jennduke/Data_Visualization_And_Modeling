2a. Are there more movies or TV shows on Netflix?

```python
content_counts = df["type"].value_counts()

print(content_counts)
```

There are more Movies!

2b. What contries produce the most Netflix content?

```python
country_counts = df["country"].value_counts()

print(country_counts.head(10))
```

The United States produces the most content.

2c. What year had the most netflix titles?

```python
year_counts = df["release_year"].value_counts()

print(year_counts.head(10))
```

In 2018 the most content was released.

2d. What is the most common genre on Netflix? 

```python
genre_counts = df["listed_in"].value_counts()

print(genre_counts.head(10))
```

Dramas, International Movies had the most in the genre cateogory. 

2e. What rating is most common for movies? 

```python
movie_ratings = df[df["type"] == "Movie"]["rating"].value_counts()

print(movie_ratings)
```

TV-MA is the most common rating for movies.
