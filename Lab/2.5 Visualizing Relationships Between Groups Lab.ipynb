# Lab Instructions

You are part of a data team for a financial institution that uses an AI model to determine who will and will not be granted a loan.  Recently, there have been complaints from customers that the model unfairly discriminates based on the applicant's age and sex. A court has ordered that the financial institution must provide information about how the model works.  

Data from recent loan applications is given in `loan_data.csv`.  The data dictionary is below:

| Feature              | Description | Data Type |
| :---------------- | :------: | ----: |
|person_age|Age of the person|Float
|person_gender|Gender of the person|Categorical
|person_education|Highest education level|Categorical
|person_income|Annual income|Float
|person_emp_exp|Years of employment experience|Integer
|person_home_ownership|Home ownership status (e.g., rent, own, mortgage)|Categorical
|loan_amnt|Loan amount requested|Float
|loan_intent|Purpose of the loan|Categorical
|loan_int_rate|Loan interest rate|Float
|loan_percent_income|Loan amount as a percentage of annual income|Float
|cb_person_cred_hist_length|Length of credit history in years|Float
|credit_score|Credit score of the person|Integer
|previous_loan_defaults_on_file|Indicator of previous loan defaults|Categorical
|loan_status (target variable)|Loan approval status: 1 = approved; 0 = rejected|Integer

Visualize the relationship between age, sex and 5 other features with `loan_status`.  Which features seem to drive the model's decision to grant a loan?  Which seem unrelated?  Should the financial institution be concerned about potential legal trouble?  Make sure to justify your answers with specific references to your visualizations.

**You must use quantitative visualizations (boxplots or histograms) when comparing quantitative features.  DO NOT categorize quantitative features.**

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('loan_data.csv')

df.head()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(
    data=df,
    x='person_age',
    hue='loan_status',
    bins=20,
    kde=True
)

plt.title("Age vs Loan Status")
plt.xlabel("Age")
plt.ylabel("Count")
plt.show()
```

```python
plt.figure(figsize=(6,5))

sns.countplot(
    data=df,
    x='person_gender',
    hue='loan_status'
)

plt.title("Gender vs Loan Status")
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    data=df,
    x='loan_status',
    y='person_income'
)

plt.title("Income vs Loan Status")
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    data=df,
    x='loan_status',
    y='credit_score'
)

plt.title("Credit Score vs Loan Status")
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    data=df,
    x='loan_status',
    y='loan_amnt'
)

plt.title("Loan Amount vs Loan Status")
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    data=df,
    x='loan_status',
    y='loan_percent_income'
)

plt.title("Loan Percent Income vs Loan Status")
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.countplot(
    data=df,
    x='person_home_ownership',
    hue='loan_status'
)

plt.title("Home Ownership vs Loan Status")
plt.xticks(rotation=20)
plt.show()
```

```python
The approval model was analyzed by visualizing the relationship between loan_status and several features including age, gender, income, credit score, loan amount, loan percent income,
and home ownership status. The end goal was to determine which features appear to influence loan approval decisions and weather the model made unfair discriminatory decisions based on age or sex. 

The histogram compared age and loan status. It showed a very wide range of age groups, both approved and rejected. Loans were not common above the age of 60.
Older age may not show as much as an approval rate. However, that visual was fairly evenly distributed. 

The count plot compares gender and loan status. There did appear to be more male applicants, but both genders appeared to have similar proportions when
it came to approval. Gender does not appear to be an influence when deciding loan approval.

When examining the boxplot, you can see that applicants with higher incomes were approved at a higher rate than those with lower incomes. This indicates that income 
was a deciding factor in loan approval.

Credit score shows the most close relationship in this analysis. Approved applicants most always had a much higher credit score than those who were rejected. 

When viewing the loan amount, the boxplot reveals higher loans are more likely to be rejected, most likely because they are a higher risk.

The loan percent income visualization reveals a strong relation to loan approval. Applicants which were approved had much lower loan to income rations than those who were denied. 

Applicants who owned a home were more favored to be approved for a loan. This is because maintaining a mortgage shows dedication and financial stability. 

Overall, age and gender did not appear to be a deciding factor in determining loan approval. Credit score, loan income percent, income, loan amount, and ownership status appear to be 
the factor that determine whether you will be approved for a loan. 
```
