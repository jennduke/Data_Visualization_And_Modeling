The dataset used here represents Student Performance in Exams. It contians information of around 1,000 students and their academic performance. This analyzes facotors that infulence student achievement. it features gender, race/ethnicity, parental level of education, lunch price, test prep course, math score, reading score, and writing score. 

For this project the Math Score, ranging from 0 to 100, will represent the target feature. The math score represents each student's performance in mathematics. 

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Makes plots appear in the notebook
%matplotlib inline

df=pd.read_csv("StudentsPerformance.csv")
df.head()
```

```python
plt.figure(figsize=(8,5))
plt.scatter(df['reading score'], df['math score'])
plt.xlabel('Reading Score')
plt.ylabel('Math Score')
plt.title('Reading Score vs Math Score')
plt.show()
```

The scatterplot shows a strong positive relationship between both reading and math scores. Most students who had higher reading scores tend to have higher math scores as well. This suggest that the two are strongly associated. 

```python
plt.figure(figsize=(8,5))
plt.scatter(df['writing score'], df['math score'])

plt.xlabel('Writing Score')
plt.ylabel('Math Score')
plt.title('Writing Score vs Math Score')

plt.show()
```

As with the reading scores, higher writing scores also show a strong correlation with higher math scores. Those who were better in writing tend to perform better in math as well. 

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    x='gender',
    y='math score',
    data=df
)

plt.title('Gender vs Math Score')

plt.show()
```

The box plot shows the relationship between gender and student's math score. Under a very slight variation, males socred higher. 

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    x='lunch',
    y='math score',
    data=df
)

plt.title('Lunch Type vs Math Score')

plt.show()
```

Students who received standard priced lunch generally performed better in math than those with free/reduced lunch. This suggects economic factors show an impact on academic achievement. 

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    x='test preparation course',
    y='math score',
    data=df
)

plt.title('Test Preparation Course vs Math Score')

plt.show()
```

Acording to the boxplot, those who participated in a test preparation course generally had performed better in Math. This suggests that additional study may improve performance. 

```python
plt.figure(figsize=(10,5))

sns.boxplot(
    x='parental level of education',
    y='math score',
    data=df
)

plt.xticks(rotation=30)

plt.title('Parental Education vs Math Score')

plt.show()
```

Students who had parents with higher education tend to have better performance than those with just general education. This suggests that parents may have an impact on a student's success. 

The features that were most related were Reading Score and Writing Score. The ones that mattered the least were gender and Parental level of education.

The visualizations conclude that academic performance is highly related across subjects. Reading and writing scores showed the strongest correlation with Math scores. This suggests, students who excel in those subjects also may perform better in Math. Other factors, such as test preperation course and lunch cost, seemed to have a high impact on the students scores as well. Parent's education and gender had a weaker corelation. Overall, academic skills as a whole as well as study prep have strongly relate to exceling in Math.
